## Question 1: Days to Seconds

````python
# Enter a number of days
days = int(input("Please enter the number of days: "))

#Get number of seconds by multiplying no.of hours in a day by minutes in an hour and seconds in a minute.
seconds = days * 24 * 60 * 60

# Display the number of seconds
print(f"{days} day(s) is equal to {seconds} seconds.")
````

## Question 2: Volume of a Sphere

````python
# Import math to access pi
import math

#Input the radius of the sphere
radius = float(input("Enter the radius of the sphere: "))

# volume of a sphere = (4/3) * pi * r^3
#r^3==r**3
volume = (4/3) * math.pi * (radius ** 3)

# Display the volume rounded off to 2 d.p
print(f"The volume of the sphere is {volume:.2f}")
````

## Question 3: Area and Perimeter of a Square (using functions)

````python
# Define function to calculate the area of a square
def calculate_area(side):
    return side * side

#Define function to calculate the perimeter
def calculate_perimeter(side):
    return 4 * side

# Request side length as user input
side_length = float(input("Enter the side length of the square: "))

# Call the functions passing in the side length
area = calculate_area(side_length)
perimeter = calculate_perimeter(side_length)

# Display the area and volume
print(f"Area of the square: {area}")
print(f"Perimeter of the square: {perimeter}")
````

## Question 4: Check if a Character is Uppercase or Lowercase (using functions)

````python
# Define function that checks whether a character is uppercase, lowercase, or neither
def check_case(character):
    # isupper() returns True if the character is an uppercase letter
    if character.isupper():
        return "uppercase"
    # islower() returns True if the character is a lowercase letter
    elif character.islower():
        return "lowercase"
    # if it's neither (e.g. a number or symbol), let the user know
    else:
        return "neither uppercase nor lowercase (not a letter)"

#Input a single character
user_char = input("Enter a character: ")

# Call the function and store result
result = check_case(user_char)

# Display the result
print(f"The character '{user_char}' is {result}.")
````

## Question 5: Pseudocode Translation
Pseudocode:
```
BEGIN
  SET x TO 0, y TO 20
  REPEAT
    SUBTRACT 4 FROM y
    ADD 2/y TO x
  UNTIL y IS LESS THAN 6
  DISPLAY x
END
```

````python
# Set X AND Y as the intial values
x = 0
y = 20

#While loop
while True:
    # Subtract 4 from y
    y = y - 4

    # Add (2 / y) to x
    x = x + (2 / y)

    # Check the stopping condition
    # If y is less than 6, we stop the loop
    if y < 6:
        break

# Display the final value of x
print(f"x = {x}")
````

## Question 6: Average of 5 User-Input Values

````python
# Create an empty list to store  values
values = []

# For loop to allow entry of 5 values
for i in range(5):
    # Request for a value each time
    num = float(input(f"Enter value {i + 1}: "))
    # Add the value to the values list
    values.append(num)

# Calculate the average
average = sum(values) / len(values)

# Display the value list and average
print(f"Values entered: {values}")
print(f"Average of the values: {average}")
````
